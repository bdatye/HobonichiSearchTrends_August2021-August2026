Adding the necessary libraries

```python
#!/usr/bin/env python3
import pandas as pd
import matplotlib.pyplot as plt
```

Loading data

```python
#getting timeing and extracting year into a separate column
HobonichiSearch_df = pd.read_csv('NecessaryFiles/Worldwide_Hobonichi_YT_Search_time_series_Worldwide_20210816-1231_20260816-1231.csv', usecols=['Time'])

#print(HobonichiSearchesdf.head(10))
HobonichiSearch_df['Time'] = pd.to_datetime(HobonichiSearch_df['Time'])
HobonichiSearch_df['Year'] = HobonichiSearch_df['Time'].dt.year
HobonichiSearch_df['Month'] = HobonichiSearch_df['Time'].dt.month



#Getting data from csvs
#Worldwide data
HobonichiSearch_df['Worldwide_Web_Search'] = pd.read_csv('NecessaryFiles/Worldwide_Hobonichi_Web_Search_time_series_Worldwide_20210816-1230_20260816-1230.csv', usecols=['hobonichi'])
HobonichiSearch_df['Worldwide_YT_Search'] = pd.read_csv('NecessaryFiles/Worldwide_Hobonichi_YT_Search_time_series_Worldwide_20210816-1231_20260816-1231.csv', usecols=['hobonichi'])

#US
HobonichiSearch_df['US_Web_Search'] = pd.read_csv('NecessaryFiles/US_Hobonichi_Web_Search_time_series_US_20210816-1145_20260816-1145.csv', usecols=['hobonichi'])
HobonichiSearch_df['US_YT_Search'] = pd.read_csv('NecessaryFiles/US_Hobonichi_YT_Search_time_series_US_20210816-1145_20260816-1145.csv', usecols=['hobonichi'])

#Japan
HobonichiSearch_df['JP_(ほぼ日)_Web_Search'] = pd.read_csv('NecessaryFiles/JP_Hobonichi_(ほぼ日)_Web_Search_time_series_JP_20210816-1238_20260816-1238.csv', usecols=['ほぼ日'])
HobonichiSearch_df['JP_(ほぼ日)_YT_Search'] = pd.read_csv('NecessaryFiles/JP_Hobonichi_(ほぼ日)_YT_Search_time_series_JP_20210816-1239_20260816-1239.csv', usecols=['ほぼ日'])
HobonichiSearch_df['JP_Hobonichi_Web_Search'] = pd.read_csv('NecessaryFiles/JP_Hobonichi_Web_Search_time_series_JP_20210816-1233_20260816-1233.csv', usecols=['hobonichi'])
HobonichiSearch_df['JP_Hobonichi_YT_Search'] = pd.read_csv('NecessaryFiles/JP_Hobonichi_YT_Search_time_series_JP_20210816-1232_20260816-1232.csv', usecols=['hobonichi'])


print(HobonichiSearch_df.head(10))
```

Exploratory Data Analysis

```python
df = HobonichiSearch_df
fig, ax = plt.subplots(figsize=(8, 5))
for columnName in df.columns:
    #if columnName != 'Time' and columnName != 'Year': 
    if 'Search' in columnName:
        #print(columnName)
        ax.plot(df['Time'], df[columnName], label=columnName)

#ax.plot(df['Time'], df['US_YT_Search'], label='US_YT_Search')
ax.legend(loc='upper left', fontsize = 'small')
ax.set_title('Hobonichi searches over time')
ax.grid(alpha = 0.3)
ax.set_xlabel('Year')
ax.set_ylabel('Number of searches')
```
